```python
from mlflow import MlflowClient

client = MlflowClient()

experiment = mlflow.get_experiment_by_name("dvc-pipeline")

logged_models = client.search_logged_models(
    experiment_ids=[experiment.experiment_id]
)

print(f"Found {len(logged_models)} logged models\n")

for lm in logged_models:
    print(lm)
    print("=" * 80)
```

```python
for lm in logged_models:
    print("Model ID      :", getattr(lm, "model_id", None))
    print("Name          :", getattr(lm, "name", None))
    print("Source Run ID :", getattr(lm, "source_run_id", None))
    print("Status        :", getattr(lm, "status", None))
    print("Artifact URI  :", getattr(lm, "artifact_uri", None))
    print()
```
